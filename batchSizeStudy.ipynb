```python
# =========================================================
# REPRODUCIBLE BATCH SIZE STUDY
# PHYSICS-GUIDED MBVD FORMULATION
# WITH REPRESENTATIVE PLOTS
# =========================================================

import os
import random
import numpy as np
import tensorflow as tf
import matplotlib.pyplot as plt

# =========================================================
# FULL REPRODUCIBILITY
# =========================================================
SEED = 0

os.environ["PYTHONHASHSEED"] = str(SEED)

random.seed(SEED)

np.random.seed(SEED)

tf.keras.utils.set_random_seed(SEED)

tf.config.experimental.enable_op_determinism()

tf.get_logger().setLevel("ERROR")

# =========================================================
# CONSTANTS
# =========================================================
PI = np.pi

N_FREQ = 200

N_SYNTH = 120

# =========================================================
# REAL RESONATORS
# =========================================================
# INPUT DESCRIPTORS:
# (Cs, fr, Qs, Qp, R0, k2)
# =========================================================
real_resonators = [

{"Cs":7.95e-13,"fr":2468e6,"Qs":887,"Qp":3770,"R0":0.257,"k2":0.080},
{"Cs":5.989e-13,"fr":2466e6,"Qs":1034,"Qp":1379,"R0":0.257,"k2":0.080},
{"Cs":3.981e-13,"fr":2468e6,"Qs":1994,"Qp":1052,"R0":0.257,"k2":0.085},
{"Cs":1.857e-12,"fr":2461e6,"Qs":935,"Qp":2513,"R0":0.257,"k2":0.081},
{"Cs":1.39e-12,"fr":2488e6,"Qs":804,"Qp":1732,"R0":0.257,"k2":0.084},
{"Cs":1.65e-12,"fr":2324e6,"Qs":1038,"Qp":1112,"R0":0.257,"k2":0.091},
{"Cs":1.70e-12,"fr":2324e6,"Qs":911,"Qp":1683,"R0":0.257,"k2":0.0855},
{"Cs":2.60e-12,"fr":1298e6,"Qs":1470,"Qp":2580,"R0":0.257,"k2":0.093},
{"Cs":2.03e-12,"fr":1487e6,"Qs":1249,"Qp":1966,"R0":0.257,"k2":0.097},
{"Cs":2.36e-12,"fr":1477e6,"Qs":904,"Qp":1153,"R0":0.257,"k2":0.096},
{"Cs":1.096e-12,"fr":1722e6,"Qs":2688,"Qp":3530,"R0":0.257,"k2":0.092},
{"Cs":3.01e-12,"fr":1443e6,"Qs":1353,"Qp":1585,"R0":0.257,"k2":0.091},
{"Cs":3.244e-12,"fr":1269e6,"Qs":1643,"Qp":1645,"R0":0.257,"k2":0.0891},
{"Cs":1.242e-12,"fr":2847e6,"Qs":906,"Qp":2170,"R0":0.257,"k2":0.0747},
{"Cs":1.592e-12,"fr":2467e6,"Qs":832,"Qp":2911,"R0":0.257,"k2":0.0846},
{"Cs":1.90e-12,"fr":2314e6,"Qs":1181,"Qp":2603,"R0":0.257,"k2":0.086},
{"Cs":2.132e-12,"fr":2074e6,"Qs":892,"Qp":3792,"R0":0.257,"k2":0.089},
{"Cs":2.47e-12,"fr":1719e6,"Qs":974,"Qp":4742,"R0":0.257,"k2":0.090},
{"Cs":1.07e-12,"fr":1723e6,"Qs":2291,"Qp":1939,"R0":0.257,"k2":0.093},
{"Cs":2.02e-12,"fr":1485e6,"Qs":1089,"Qp":6853,"R0":0.257,"k2":0.094},
{"Cs":1.57e-12,"fr":2467e6,"Qs":1095,"Qp":1312,"R0":0.257,"k2":0.0868},
{"Cs":1.58e-12,"fr":2092e6,"Qs":857,"Qp":1712,"R0":0.257,"k2":0.092}

]

# =========================================================
# SYNTHETIC RESONATOR GENERATION
# =========================================================
def generate_synthetic_resonators(n):

    resonators = []

    while len(resonators) < n:

        resonators.append({

            "Cs": np.random.uniform(0.5e-12,3.5e-12),

            "fr": np.random.uniform(1.2e9,2.9e9),

            "Qs": np.random.uniform(700,3000),

            "Qp": np.random.uniform(1000,7000),

            "R0": 0.257,

            "k2": np.random.uniform(0.074,0.10)

        })

    return resonators

# =========================================================
# PHYSICS-GUIDED MBVD RECONSTRUCTION
# =========================================================
def compute_mbvd_parameters(r):

    Cs = r["Cs"]

    fr = r["fr"]

    Qs = r["Qs"]

    Qp = r["Qp"]

    R0 = r["R0"]

    k2 = r["k2"]

    # =====================================================
    # ANTI-RESONANCE FREQUENCY
    # =====================================================
    fa = (

        2*fr

        /

        (

            1 +

            np.sqrt(

                1 - (16*k2)/(PI**2)

            )

        )

    )

    wr = 2*PI*fr

    wa = 2*PI*fa

    # =====================================================
    # MOTIONAL CAPACITANCE
    # =====================================================
    Cm = Cs*((fa/fr)**2 -1)

    # =====================================================
    # MOTIONAL INDUCTANCE
    # =====================================================
    Lm = 1/((wr**2)*Cm)

    # =====================================================
    # MOTIONAL RESISTANCE
    # =====================================================
    Rm = wr*(Lm/Qs) - R0

    # =====================================================
    # STATIC BRANCH LOSS
    # =====================================================
    Rs = wa*(Lm/Qp) - Rm

    # =====================================================
    # SAFETY
    # =====================================================
    Rm = np.maximum(Rm,1e-6)

    Rs = np.maximum(Rs,1e-6)

    Cm = np.maximum(Cm,1e-18)

    Lm = np.maximum(Lm,1e-18)

    return Rm,Lm,Cm,Cs,R0,Rs,fa

# =========================================================
# MBVD RESPONSE
# =========================================================
def compute_Y11(r):

    fr = r["fr"]

    Rm,Lm,Cm,Cs,R0,Rs,fa = compute_mbvd_parameters(r)

    f = np.linspace(

        fr*0.93,

        fa*1.07,

        N_FREQ

    )

    w = 2*PI*f

    # =====================================================
    # STATIC BRANCH
    # =====================================================
    Zs = R0 + 1/(1j*w*Cs)

    # =====================================================
    # MOTIONAL BRANCH
    # =====================================================
    Zm = Rm + 1/(1j*w*Cm) + 1j*w*Lm

    # =====================================================
    # PARALLEL COMBINATION
    # =====================================================
    Zp = 1/(1/Zs + 1/Zm)

    # =====================================================
    # TOTAL IMPEDANCE
    # =====================================================
    Ztotal = Rs + Zp

    # =====================================================
    # ADMITTANCE
    # =====================================================
    Y11 = np.abs(1/Ztotal)

    return Y11,f

# =========================================================
# PRECOMPUTE CURVES
# =========================================================
synthetic_resonators = generate_synthetic_resonators(N_SYNTH)

all_resonators = real_resonators + synthetic_resonators

precomputed_data = {}

for r in all_resonators:

    Y,f = compute_Y11(r)

    precomputed_data[id(r)] = (Y,f)

# =========================================================
# FREQUENCY EXTRACTION
# =========================================================
def extract_frequencies(Y,f):

    fr = f[np.argmax(Y)]

    fa = f[np.argmin(Y)]

    return fr,fa

# =========================================================
# FINAL STANDARDIZED MLP
# =========================================================
def build_model():

    tf.keras.backend.clear_session()

    model = tf.keras.Sequential([

        tf.keras.layers.Input((6,)),

        tf.keras.layers.Dense(
            96,
            activation="swish"
        ),

        tf.keras.layers.Dense(
            96,
            activation="swish"
        ),

        tf.keras.layers.Dense(
            128,
            activation="swish"
        ),

        tf.keras.layers.Dense(
            N_FREQ,
            activation="softplus"
        )
    ])

    model.compile(

        optimizer=tf.keras.optimizers.Adam(
            learning_rate=1e-3
        ),

        loss="mse"
    )

    return model

# =========================================================
# BATCH SIZES
# =========================================================
batch_sizes = [8,16,32,64]

results = {}

example_predictions = {}

# =========================================================
# MAIN EXPERIMENT LOOP
# =========================================================
for batch in batch_sizes:

    print(f"\nTesting batch size: {batch}")

    mse_list = []

    res_error = []

    anti_error = []

    representative_example = None

    # =====================================================
    # LOOCV
    # =====================================================
    for i in range(len(real_resonators)):

        test_res = real_resonators[i]

        train_real = (

            real_resonators[:i]

            +

            real_resonators[i+1:]

        )

        train_res = train_real + synthetic_resonators

        X_train = []

        Y_train = []

        # =================================================
        # BUILD TRAINING DATA
        # =================================================
        for r in train_res:

            Rm,Lm,Cm,Cs,R0,Rs,fa = compute_mbvd_parameters(r)

            X_train.append([

                Rm,
                Lm,
                Cm,
                Cs,
                R0,
                Rs

            ])

            Y_train.append(

                precomputed_data[id(r)][0]

            )

        X_train = np.array(X_train,np.float32)

        Y_train = np.array(Y_train,np.float32)

        # =================================================
        # NORMALIZATION
        # =================================================
        Xm = X_train.mean(axis=0)

        Xs = X_train.std(axis=0)

        Ym = Y_train.mean(axis=0)

        Ys = Y_train.std(axis=0)

        Xs = np.where(Xs < 1e-12,1.0,Xs)

        Ys = np.where(Ys < 1e-12,1.0,Ys)

        Xn = (X_train - Xm)/Xs

        Yn = (Y_train - Ym)/Ys

        # =================================================
        # MODEL
        # =================================================
        model = build_model()

        earlystop = tf.keras.callbacks.EarlyStopping(

            monitor="loss",

            patience=15,

            restore_best_weights=True

        )

        # =================================================
        # TRAINING
        # =================================================
        model.fit(

            Xn,
            Yn,

            epochs=120,

            batch_size=batch,

            shuffle=True,

            verbose=0,

            callbacks=[earlystop]

        )

        # =================================================
        # TEST SAMPLE
        # =================================================
        Rm,Lm,Cm,Cs,R0,Rs,fa = compute_mbvd_parameters(test_res)

        Xt = np.array([[

            Rm,
            Lm,
            Cm,
            Cs,
            R0,
            Rs

        ]],dtype=np.float32)

        Xt = (Xt - Xm)/Xs

        # =================================================
        # PREDICTION
        # =================================================
        Y_pred = model.predict(

            Xt,
            verbose=0

        )[0]

        Y_pred = Y_pred*Ys + Ym

        Y_true,f = precomputed_data[id(test_res)]

        # =================================================
        # METRICS
        # =================================================
        mse_list.append(

            np.mean((Y_true - Y_pred)**2)

        )

        fr_true,fa_true = extract_frequencies(Y_true,f)

        fr_pred,fa_pred = extract_frequencies(Y_pred,f)

        res_error.append(

            abs(fr_true - fr_pred)/1e6

        )

        anti_error.append(

            abs(fa_true - fa_pred)/1e6

        )

        # =================================================
        # REPRESENTATIVE PLOT SAMPLE
        # =================================================
        if i == 0:

            representative_example = {

                "f":f,

                "Y_true":Y_true,

                "Y_pred":Y_pred

            }

    # =====================================================
    # STORE RESULTS
    # =====================================================
    results[batch] = {

        "MSE":np.mean(mse_list),

        "ResError":np.mean(res_error),

        "AntiError":np.mean(anti_error)

    }

    example_predictions[batch] = representative_example

# =========================================================
# RESULTS TABLE
# =========================================================
print("\n================ BATCH SIZE STUDY ================")

print(f"{'Batch':<10}"
      f"{'Curve MSE':<20}"
      f"{'Res Error (MHz)':<20}"
      f"{'Anti Error (MHz)':<20}")

for b,v in results.items():

    print(f"{b:<10}"
          f"{v['MSE']:<20.6e}"
          f"{v['ResError']:<20.4f}"
          f"{v['AntiError']:<20.4f}")

print("==================================================")

# =========================================================
# REPRESENTATIVE CURVE PLOTS
# =========================================================
for batch, ex in example_predictions.items():

    plt.figure(figsize=(8,5))

    plt.plot(

        ex["f"]/1e6,

        20*np.log10(ex["Y_true"] + 1e-15),

        linewidth=3,

        label="Ground Truth"

    )

    plt.plot(

        ex["f"]/1e6,

        20*np.log10(ex["Y_pred"] + 1e-15),

        "--",

        linewidth=2,

        label=f"Batch {batch}"

    )

    plt.xlabel("Frequency (MHz)")

    plt.ylabel("|Y11| (dB)")

    plt.title(

        f"Batch Size {batch}: Representative Admittance Reconstruction"

    )

    plt.grid(True)

    plt.legend()

    plt.tight_layout()

    plt.show()

# =========================================================
# BAR PLOTS
# =========================================================
batch_list = sorted(results.keys())

mse_values = [results[b]["MSE"] for b in batch_list]

res_values = [results[b]["ResError"] for b in batch_list]

anti_values = [results[b]["AntiError"] for b in batch_list]

# =========================================================
# CURVE MSE
# =========================================================
plt.figure(figsize=(6,4))

plt.bar(batch_list,mse_values)

plt.xlabel("Batch Size")

plt.ylabel("Mean Curve MSE")

plt.title("Batch Size Comparison: Curve Reconstruction Error")

plt.grid(True)

plt.tight_layout()

plt.show()

# =========================================================
# RESONANCE ERROR
# =========================================================
plt.figure(figsize=(6,4))

plt.bar(batch_list,res_values)

plt.xlabel("Batch Size")

plt.ylabel("Mean Resonance Error (MHz)")

plt.title("Batch Size Comparison: Resonance Frequency Error")

plt.grid(True)

plt.tight_layout()

plt.show()

# =========================================================
# ANTI-RESONANCE ERROR
# =========================================================
plt.figure(figsize=(6,4))

plt.bar(batch_list,anti_values)

plt.xlabel("Batch Size")

plt.ylabel("Mean Anti-Resonance Error (MHz)")

plt.title("Batch Size Comparison: Anti-Resonance Frequency Error")

plt.grid(True)

plt.tight_layout()

plt.show()
```
